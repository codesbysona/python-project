```python
#student performance operator analyzer
#student details
name='Sona Singh'
print(name)
print('subject____marks')
maths    =   70
physics  =   50
chemistry=   65
english  =   85
PE       =   90
```

```python
#Arithmetic operators
total=maths+physics+chemistry+english+PE
average=total/5
print('total marks=',total)
print('average=',average)
#here we used addition and division
#EXAMPLE OF OTHER ARITHMETIC OPERATORS
#subtraction
a=10
b=5
print(a-b)
print(a*b)
print(a**b)
print(a//b)
print(a%b)
```

```python
#Comparison operators
print('maths=70',maths==70)
print('physics not = 50',physics!=50)
print('chemistry>=physics',chemistry>=physics)
print('pe>english',PE>english)
print('english<=maths',english<=maths)
```

```python
#Logical operators
passed=maths>=40 and physics>=50
fail=chemistry>=70 or maths<=40
good_performance=english>=80 and PE>=80
need_improvement=not(maths<=50)
print('passed=',passed)
print('fail=',fail)
print('the student had an excellent performce in english and pe=',good_performance)
print('the student needs improvement in physics=',need_improvement)
```

```python
#assignment operator
bonus=5
bonus+=2
print('after+=',bonus)
bonus-=1
print('after-=',bonus)
bonus*=2
print('after*=',bonus)
```

```python
#membership operators
subjects='maths','physics','chemistry','english','PE'
print('maths' in subjects)
print('biology' in subjects)
print('biology' not in subjects)
```

```python
#identity operators
subjects1='maths','physics','chemistry'
subjects2='english','PE'
subjects3=subjects2
print('main subjects are not PCM=',subjects1 is  subjects2)
print('additional subjects are PE and english=',subjects3 is subjects2)

```

```python
#bitwise operators
#here we will use bitwise operator for marks of MATHS and PHYSICS
print('AND=',maths & physics)
print('OR=',maths | physics)
print('XOR=',maths ^ physics)
print('left shift=',maths<<1)
print('right shift=',physics>>2)
```

```python
#ternary operators
result='PASS' if total>=300 else 'FAIL'
print('RESULT=',result)
```

```python
#operator precedence
x=maths+physics*english
y=(maths+physics)*english
z=PE*(chemistry/english+maths)//physics
print('order of operators')
print('multiplication>addition=',x)
print('parenthesis>multiplication=',y)
print('parenthesis>multiplication>floor division=',z)
```
